## M06 准备：输出文件怎样保存

`RUN` 是本次新建的成果目录，`ROOT` 是原始材料目录。下面三个保存工具只负责把表、文字和图片写入文件，不替你进行科学分析。后续每个保存调用都会给出文件名；打开打印出的 RUN 目录即可找到。重新运行准备单元会另建目录，因此要记下本次路径。

````python
from pathlib import Path
from datetime import datetime, timezone
import csv, json, uuid, os, hashlib, copy
import numpy as np
import matplotlib.pyplot as plt
import lightkurve as lk
from astropy.io import fits
from IPython.display import display, Image
from analyze import CONFIG, ROOT, plain

# Each execution starts an independent persistent folder. Existing runs remain.
RUN = Path(os.environ.get('SYSTEMEDU_GUIDE_OUTPUT',
    str(ROOT / 'student-runs' / (datetime.now(timezone.utc).strftime('%Y%m%dT%H%M%S') + '-' + uuid.uuid4().hex[:8]))))
RUN.mkdir(parents=True, exist_ok=False)
def output_path(name):
    path = (RUN / name).resolve()
    if not path.is_relative_to(RUN.resolve()):
        raise ValueError('Output must stay in this run folder')
    path.parent.mkdir(parents=True, exist_ok=True)
    return path
def save_json(name, value):
    path = output_path(name)
    path.write_text(json.dumps(plain(value), ensure_ascii=False, indent=2, allow_nan=False) + '\n')
    return path
def save_csv(name, headers, rows):
    path = output_path(name)
    with path.open('w', newline='') as stream:
        writer = csv.writer(stream)
        writer.writerow(headers)
        for row in rows:
            writer.writerow([row.get(key, '') for key in headers] if isinstance(row, dict) else row)
    return path
def save_plot(name):
    path = output_path(name)
    plt.gcf().savefig(path, dpi=150, bbox_inches='tight')
    return path
def load_json(name):
    return json.loads((RUN / name).read_text())
print('原始材料目录:', ROOT)
print('本次成果目录（请记下）:', RUN)
assert (ROOT / 'raw').is_dir(), 'Notebook must be beside analyze.py and raw/'
student_input = ROOT / 'student-input'
student_input.mkdir(exist_ok=True)
window_input = student_input / 'candidate-window.json'
if not window_input.exists():
    window_input.write_text(json.dumps({'start_bkjd':None,'end_bkjd':None,
        'boundary':'[start,end)','cadenceno':[],'observation':''},ensure_ascii=False,indent=2))
````

### M07 · 建立一列质量判断
先用构造数 5&4、5&2 预测结果 4、0，再执行真实规则。qkeep 不是删过的表，而是和原表等长的一列真假。np.unique 返回整数值与次数两列，zip 逐对取出；循环缩进部分对每对执行一次。

````python
print('构造编码：', 5 & 4, 5 & 2)
q = np.asarray(d['SAP_QUALITY'], dtype=np.int64)
reject_bits = 1130799
assert lk.KeplerQualityFlags.DEFAULT_BITMASK == reject_bits, '库版本/默认规则不同，请核对环境，不改原数据。'
qkeep = (q & reject_bits) == 0
values, counts = np.unique(q, return_counts=True)
quality_rows = []
print('质量整数 / 条数 / 当前质量关通过 / 状态名称')
for value, count in zip(values, counts):
    names = lk.KeplerQualityFlags.decode(int(value))
    passed = (int(value) & reject_bits) == 0
    print(int(value), int(count), passed, '; '.join(names) or '无已记录状态位')
    quality_rows.append([int(value), int(count), bool(passed), '; '.join(names)])
quality_rejected = int((~qkeep).sum())
print('原表行数 / 质量拒绝 / 质量通过', len(d), quality_rejected, int(qkeep.sum()))
assert len(qkeep) == len(d) and quality_rejected == 230
save_csv('quality-counts.csv', ['quality', 'rows', 'passes_quality_rule', 'decoded'], quality_rows)
````

### M07 · 用编号查一条通过和一条拒绝
flatnonzero 返回条件为真的位置；[0] 取其中第一个位置，然后用同一个位置读取编号和 q。这里只选示范记录，不改变规则。请记下两条编号，并解释“非零但通过”为什么可能。下一节继续检查它们的数值。

````python
quality_examples = []
for label, condition in [('nonzero_quality_pass', (q != 0) & qkeep), ('quality_rejected', ~qkeep)]:
    j = int(np.flatnonzero(condition)[0])
    example = {'kind': label, 'raw_row': j, 'cadenceno': int(d['CADENCENO'][j]),
               'quality': int(q[j]), 'quality_keep': bool(qkeep[j]),
               'decoded': lk.KeplerQualityFlags.decode(int(q[j]))}
    print(example)
    quality_examples.append(example)
save_json('quality-rule.json', {'lightkurve_version': lk.__version__, 'reject_bits': reject_bits,
    'rule': '(SAP_QUALITY & 1130799) == 0', 'raw_rows': len(d),
    'quality_rejected': quality_rejected, 'examples': quality_examples})
````

### M08 · 停靠 1：先手判七条构造记录
下面不是 Kepler 数据。先预测 A–G 哪些通过数值条件：时间/通量/误差都有限，且误差>0。注意 C 的通量为 0，D 的误差为 0，它们不是同一条件。先写自己的真假，再运行对照。

数组同一位置属于同一行；& 连接同一行必须同时成立的条件。finite_demo 只表示这组数值条件，不说明测量绝对可靠。

````python
demo_id = ['A', 'B', 'C', 'D', 'E', 'F', 'G']
demo_time = np.array([0, 1, 2, 3, 4, np.nan, 6], dtype=float)
demo_flux = np.array([100, np.nan, 0, 100, 100, 100, np.inf], dtype=float)
demo_error = np.array([1, 1, 1, 0, -1, 1, 1], dtype=float)
demo_time_ok = np.isfinite(demo_time)
demo_flux_ok = np.isfinite(demo_flux)
demo_error_ok = np.isfinite(demo_error)
demo_error_positive = demo_error > 0
finite_demo = demo_time_ok & demo_flux_ok & demo_error_ok & demo_error_positive
print('编号 时间 通量 误差 | 时间有限 通量有限 误差有限 误差正值 | 数值通过')
for j in range(len(demo_id)):
    print(demo_id[j], demo_time[j], demo_flux[j], demo_error[j], '|',
          demo_time_ok[j], demo_flux_ok[j], demo_error_ok[j], demo_error_positive[j], '|', finite_demo[j])
assert finite_demo.tolist() == [True, False, True, False, False, False, False]
````

### M08 · 停靠 2：把质量与数值两列排在一起
构造质量条件让 B 失败，其余通过。B 同时缺通量，只扣一次；D/E/F/G 是质量通过后新增数值失败，A/C 最终保留。先算 7−1−4=2，再看程序。~ 把真与假互换，sum 在布尔列里数真有几个。

````python
qkeep_demo = np.array([True, False, True, True, True, True, True])
keep_demo = qkeep_demo & finite_demo
for j in range(len(demo_id)):
    print(demo_id[j], '质量通过', bool(qkeep_demo[j]), '数值通过', bool(finite_demo[j]), '最后保留', bool(keep_demo[j]))
print('三组互不重叠数量：', int((~qkeep_demo).sum()), int((qkeep_demo & ~finite_demo).sum()), int(keep_demo.sum()))
print('全表数值失败数（不能再次整项扣除）：', int((~finite_demo).sum()))
assert int((~qkeep_demo).sum()) + int((qkeep_demo & ~finite_demo).sum()) + int(keep_demo.sum()) == 7
````

### M08 · 停靠 3：逐列检查真实 Q3，显示真实编号
下面才是 d 中的实测表。四列条件分别保留，之后合并；qkeep 仍是 M07 的结果。先检查 7404、7405、7465 的中间状态，再查看“质量通过后新增数值排除”的完整列表。每条显示自己的原因，不靠总数猜。

````python
probe_lc = lc[lc.time.value <= lc.time.value.min()+6.0].copy()
injection_centre = float(probe_lc.time.value.min()+2.0)
probe_offset = np.abs(probe_lc.time.value-injection_centre)
injection_mask = probe_offset < 0.08
background_mask = (probe_offset > 0.16) & (probe_offset < 0.7)
injected_lc = probe_lc.copy()
injected_lc.flux[injection_mask] *= (1-800e-6)
print('仅修改这些采样编号:',list(map(int,probe_lc.cadenceno[injection_mask])))
def injection_depth_ppm(curve):
    values=np.asarray(curve.flux.value)
    if injection_mask.sum()<3 or background_mask.sum()<3:
        raise ValueError('本例内窗或背景少于3点，暂不计算深度')
    outer=float(np.mean(values[background_mask]))
    inner=float(np.mean(values[injection_mask]))
    if outer<=0: raise ValueError('背景必须为正')
    return (outer-inner)/outer*1e6
recovery_rows=[]
for width in [0.15,0.5,1.0]:
    try:
        reference_test,_=local_median(probe_lc,window_day=width)
        injected_test,_=local_median(injected_lc,window_day=width)
        depth_before=injection_depth_ppm(reference_test)
        depth_after=injection_depth_ppm(injected_test)
        recovery_rows.append([width,'computed',depth_before,depth_after,depth_after-depth_before,''])
    except ValueError as error:
        recovery_rows.append([width,'insufficient_support',None,None,None,str(error)])
for row in recovery_rows: print(row)
save_csv('M12/trend-check.csv',['window_day','status','original_depth_ppm','injected_depth_ppm','recovered_depth_ppm','reason'],recovery_rows)
save_json('M12/injection-conditions.json',{'data_kind':'artificially modified copy of Q3','centre_bkjd':injection_centre,'duration_day':0.16,'input_depth_ppm':800,'background_offset_day':[0.16,0.7],'formal_curve_unmodified':bool(np.array_equal(original_flux_before_M12,lc.flux.value))})
assert np.array_equal(original_flux_before_M12,lc.flux.value)
````

## 从一张分组表开始，再测自己的候选

打开教程 M13。先处理六行构造记录：时间是相对中心的天数，不是 Kepler 的 BKJD；中间两条 97、99 在内部，另外四条 99、101、100、100 作背景。内部均值98、背景均值100，相对深度应为2%。我们先打印每条记录属于哪一组，再计算，这样能够发现选区错误。

四个边界必须满足 `left < a < b < right`。左背景从 left 到 a，内部从 a 到 b，右背景从 b 到 right；各段都含左不含右。代码中的两个比较先分别产生真假数组，再用 `&` 表示同时满足、`|` 表示满足其中之一。它们是在对每一行分类，不是在把两个通量做按位运算。

本节的局部相对深度与参考 `analyze.measure` 不同：后者按周期合并多个事件，用归一化尺度的均值差作诊断，还留出不同背景间隔。不要直接拿它代替这里的四边界计算。下一节需要事件中心：开始落在10.00–10.02天、结束落在10.10–10.12天时，中心范围是10.05–10.07天，时长范围0.08–0.12天。这只是构造例的采样限制；真实边界看不清时可以留空，不能靠更多小数位补精度。


````python
def local_relative_depth(time_values,flux_values,left,a,b,right):
    if not left<a<b<right: raise ValueError('需要 left < a < b < right；先检查四条时间边界')
    time_values=np.asarray(time_values,dtype=float)
    flux_values=np.asarray(flux_values,dtype=float)
    if time_values.shape!=flux_values.shape: raise ValueError('时间和通量必须逐行对应')
    valid_values=np.isfinite(time_values)&np.isfinite(flux_values)
    inside=(time_values>=a)&(time_values<b)&valid_values
    outside=(((time_values>=left)&(time_values<a))|((time_values>=b)&(time_values<right)))&valid_values
    if inside.sum()<2 or outside.sum()<2: raise ValueError('内部或背景少于2点；不返回零深度')
    f_in=float(np.mean(flux_values[inside]));f_out=float(np.mean(flux_values[outside]))
    if f_out<=0: raise ValueError('背景均值必须为正')
    return {'baseline':f_out,'inside_mean':f_in,'relative_depth_ppm':(f_out-f_in)/f_out*1e6,
            'n_in':int(inside.sum()),'n_out':int(outside.sum()),'minimum':float(np.min(flux_values[inside]))},inside,outside
demo_time=np.array([-0.3,-0.2,-0.1,0.1,0.2,0.3])
demo_flux=np.array([99.,101.,97.,99.,100.,100.])
demo_measure,demo_inside,demo_outside=local_relative_depth(demo_time,demo_flux,-0.4,-0.15,0.15,0.4)
demo_rows=[[float(x),float(y),bool(i),bool(o)] for x,y,i,o in zip(demo_time,demo_flux,demo_inside,demo_outside)]
print('构造例：相对天数、通量、内部、背景');print(*demo_rows,sep='\n');print(demo_measure)
assert np.isclose(demo_measure['relative_depth_ppm'],20000.)
save_csv('M13/constructed-grouping.csv',['relative_day','constructed_flux','inside','outside'],demo_rows)
fig,ax=plt.subplots(figsize=(8,3));ax.scatter(demo_time[demo_outside],demo_flux[demo_outside],label='Background')
ax.scatter(demo_time[demo_inside],demo_flux[demo_inside],label='Inside');ax.axvspan(-0.15,0.15,alpha=.15)
ax.set(xlabel='Constructed relative time [day]',ylabel='Constructed flux');ax.legend();fig.tight_layout();save_plot('M13/constructed-groups.png');plt.close(fig)
````

先把构造例中两个内窗数改为99、101，其他不变，预测深度应该是多少，再运行核对。背景与内部均值相同应为零；若内部更亮则应为负，不能取绝对值。正式选区的参数和测量另外保存。

````python
def box_score(y,inside):
    y=np.asarray(y,dtype=float);inside=np.asarray(inside,dtype=bool)
    if y.shape!=inside.shape: raise ValueError('通量与内外标记长度必须相同')
    if not np.all(np.isfinite(y)) or inside.sum()<2 or (~inside).sum()<2: raise ValueError('输入无效或某组少于2点')
    constant=float(np.mean(y));inner=float(np.mean(y[inside]));outer=float(np.mean(y[~inside]))
    model=np.where(inside,inner,outer)
    residual=y-model
    sse_constant=float(np.sum((y-constant)**2));sse_box=float(np.sum(residual**2))
    return {'inside_mean':inner,'outside_mean':outer,'sse_constant':sse_constant,'sse_box':sse_box,'delta_sse':sse_constant-sse_box},model
score_demo=np.array([10.,10.,9.,9.,10.,10.]);score_inside=np.array([False,False,True,True,False,False])
score_result,score_model=box_score(score_demo,score_inside)
score_detail=[[j,float(score_demo[j]),bool(score_inside[j]),float(score_model[j]),float(score_demo[j]-score_model[j]),float((score_demo[j]-score_model[j])**2),float((score_demo[j]-score_demo.mean())**2)] for j in range(6)]
print('row,y,inside,model,residual,residual_squared,constant_residual_squared');print(*score_detail,sep='\n');print(score_result)
assert np.isclose(score_result['delta_sse'],4/3)
save_csv('M16/constructed-score-detail.csv',['row','constructed_flux','inside','model','residual','residual_squared','constant_residual_squared'],score_detail)
errors=np.array([1.,2.]);values=np.array([1.,3.]);weights=1/errors**2
weighted_mean=float(np.sum(weights*values)/np.sum(weights))
print('构造加权：误差',errors,'权重',weights,'等权均值',values.mean(),'加权均值',weighted_mean)
assert np.isclose(weighted_mean,1.4)
score_rows=[]
if student_trial is not None:
    pp=float(student_trial['period_day']);tt=float(student_trial['epoch_bkjd'])
    phase_days=((q3_flat_initial.time.value-tt+pp/2)%pp)-pp/2
    for duration in [0.06,0.10,0.14]:
        if duration>=pp:
            score_rows.append([duration,'invalid_duration',None,None,None,None,None]);continue
        inside=np.abs(phase_days)<duration/2
        try:
            result,_=box_score(q3_flat_initial.flux.value,inside)
            score_rows.append([duration,'computed',result['inside_mean'],result['outside_mean'],result['sse_constant'],result['sse_box'],result['delta_sse']])
        except ValueError:
            score_rows.append([duration,'insufficient_data',None,None,None,None,None])
save_csv('M16/box-score-handcheck.csv',['duration_day','status','inside_mean','outside_mean','sse_constant','sse_box','delta_sse'],score_rows)
save_json('M16/score-status.json',{'real_status':'trial_computed' if student_trial is not None else 'real_trial_not_started','constructed_score':score_result,'weighted_example':weighted_mean})
````

## M24 · 在查看 Q4 亮度前预测下降时间

## 用一条真实参数，走完一行预测

先打开上一节打印的 method-freeze.json，找到 ephemeris。代码直接读取里面的完整数字，不让你从屏幕抄一个已四舍五入的周期。再从 manifest.json 找 Q4 的 lightcurve 条目，只取 fits_header 中的 TSTART 和 TSTOP。这两个数说明文件记录的起止范围，不说明中间每一时刻都有有效观测。

一行预测需要四步：选整数编号 n；算中心 t0+nP；把中心减 D/2 得开始；加 D/2 得结束。代码循环只是把同样四步对每个编号做一遍。第一段输出全部事件到 predictions.csv；请用首行、中央一行、末行在纸上分别代入核对。差一点的小数显示不影响保存的完整精度。

预测图只有时间和色条，尚无 Q4 亮度。横向色条表示名义事件区间；两条边线表示文件头的覆盖范围。边界表还检查中心在范围外、但区间与范围相交的情况，并说明主表是否收录。没有额外边界事件也保留只有表头的边界表，说明已检查而非忘记提交。

先判断构造例：范围 105–112 天、中心 112 天、D=0.2 天，中心在范围内，但结束到 112.1 天；这叫名义窗口超出覆盖边界，并不能说实际观测了完整下降。现在查看自己的末行，用同一逻辑解释，不能把构造例数字复制成你的观测。

代码把原预测 JSON 的校验值与配置关联。之后看到 Q4 低点偏到一边，也不得把这张表移动到低点处；新想法另存成探索。calculation.md 中留着你的三行验算位置。下节带着这张表打开 Q4，比较的是事先保存的时段。

````python
frozen = read_json(RUN/'audit/method-freeze.json')
e = frozen['ephemeris']
manifest = read_json(ROOT/'manifest.json')
header = next(f['fits_header'] for f in manifest['files'] if f['role']=='holdout' and f['kind']=='lightcurve')
a,b = float(header['TSTART']),float(header['TSTOP'])
P,t0,D = e['period_day'],e['epoch_bkjd'],e['duration_day']
n_first = int(np.ceil((a-t0)/P))
n_last = int(np.floor((b-t0)/P))
forecast=[]
for n in range(n_first,n_last+1):
    centre=t0+n*P
    forecast.append({'cycle':n,'centre_bkjd':centre,'start_bkjd':centre-D/2,'end_bkjd':centre+D/2})
assert forecast,'预测为空：核对时间基准和覆盖，先不读取 Q4 亮度。'
fsha=digest(RUN/'audit/method-freeze.json')
prediction_obj={'freeze_sha256':fsha,'based_on_header_only':True,'windows':forecast}
prediction_path=RUN/'audit/holdout-predictions-before-reveal.json'
if prediction_path.exists():
    assert read_json(prediction_path)==prediction_obj,'已有预测不同；保留原件并检查配置。'
else:
    save_json('audit/holdout-predictions-before-reveal.json',prediction_obj)
save_csv('M24-predictions-v1/predictions.csv',
    ['cycle','centre_bkjd','start_bkjd','end_bkjd','freeze_sha256'],
    [dict(row,freeze_sha256=fsha) for row in forecast])
edge_rows=[]
for n in range(int(np.ceil((a-D/2-t0)/P)),int(np.floor((b+D/2-t0)/P))+1):
    centre=t0+n*P
    start,end=centre-D/2,centre+D/2
    if start<a or end>b or not a<=centre<=b:
        edge_rows.append({'cycle':n,'centre_bkjd':centre,'start_bkjd':start,'end_bkjd':end,
            'in_primary':a<=centre<=b,'overlap_start_bkjd':max(a,start),'overlap_end_bkjd':min(b,end),
            'note':'nominal window intersects a header boundary; actual sampling not yet inspected'})
save_csv('M24-predictions-v1/edge-review.csv',
    ['cycle','centre_bkjd','start_bkjd','end_bkjd','in_primary','overlap_start_bkjd','overlap_end_bkjd','note'],edge_rows)
fig,ax=plt.subplots(figsize=(11,3))
for row in forecast:
    ax.axvspan(row['start_bkjd'],row['end_bkjd'],color='#418eb8',alpha=.7)
ax.axvline(a,color='black',ls='--');ax.axvline(b,color='black',ls='--')
ax.set(xlim=(a-D,b+D),ylim=(0,1),yticks=[],xlabel='BKJD (day)',title='Predictions only; no Q4 flux loaded')
fig.tight_layout();save_plot('M24-predictions-v1/predictions-only.png');plt.close(fig)
keep_text('M24-predictions-v1/calculation.md', '# 本人的预测验算\n\n[待填写] 首行：编号、完整参数代入式、开始和结束。\n\n[待填写] 中间一行：\n\n[待填写] 末行及覆盖边界说明：\n\n这些参数来自 audit/method-freeze.json，时间为 BKJD 天。\n')
for index in sorted({0,len(forecast)//2,len(forecast)-1}):
    print('待手算核对：',forecast[index])
assert all(np.isclose(y['centre_bkjd']-x['centre_bkjd'],P,rtol=0,atol=1e-10) for x,y in zip(forecast,forecast[1:]))
print('已保存',len(forecast),'行预测；边界表',len(edge_rows),'行。本节没有读取 Q4 亮度。')
````

## M26 · 比较预测与 Q4 观测的差异

## 先把每个预测留下，再谈是否相符

这节的第一张表从 M24 的原预测开始，一行也不删除。程序逐行寻找 Q4 记录，而不是只留下看起来成功的区间。窗内点数为 0 表示没有有效测量落进窗口，不是测得深度为 0。

看 coverage_status 一列前，先读它的定义。本课的自动规则只检查时间采样：窗内至少两个点、窗口前后都有近邻点、相邻有效点间隔不超过典型间隔的 1.5 倍。这是用于描述数据缺口的操作规则，不是足以检出行星的物理标准。满足时写“跨窗连续采样；信号可测性待判断”；不满足则说明哪种采样不足。即使这一列满足，几个散布很大的点也可能测不准浅下降。

每一行还有原始窗内点数、保留点数、前后点数、最大间隔及对应局部图。先从首行走一遍：找到两条窗口边线；数一数窗内点；查看周围有没有断档；把 CSV 数量与图相对照。原始有点而有效点减少时，要回查筛选，不能一律叫望远镜没观测。你的证据判断写在 learner_judgement 中；程序不替你填“支持”。

第二份表聚合整段结果。它取 Q3 和首次 Q4 的固定盒深度，按 Q4−Q3 相减。例如 100−120=−20 ppm，表示后者测得较浅，不表示下降幅度本身是负数。实际数字从两份 JSON 的 fixed_box_depth.depth_ppm 读取，并在 aggregate.json 中保留来源字段。经验标准误与两者合成尺度一起列出；这些描述量未完整考虑相关噪声和趋势不确定性，不能当发现置信度。

整体移位诊断另列分钟和天。它只问把整组预测一起平移时描述性指标如何变化；没有逐事件测出实际中心，因此不要填 O−C。请任选一行讲清采样事实，再引用整体测量解释你目前能支持到哪一步。这个解释会成为 M27 的材料。

````python
windows_rows=[]
if not Q4_REVEAL or not (RUN/'audit/holdout-first-results.json').exists():
    print('M26 待完成：尚无已获准读取的 Q4 首次结果。')
else:
    frozen=read_json(RUN/'audit/method-freeze.json')
    forecasts=read_json(RUN/'audit/holdout-predictions-before-reveal.json')['windows']
    first_holdout=read_json(RUN/'audit/holdout-first-results.json')
    t=np.asarray(holdout_flat.time.value)
    raw_t=np.asarray(holdout_bundle['data']['TIME'],dtype=float)
    cadence=float(np.median(np.diff(t)))
    rule=frozen['course_coverage_rule']
    given_counts={row['cycle']:row['accepted_observations_inside'] for row in first_holdout['predicted_windows']}
    for w in forecasts:
        start,end=w['start_bkjd'],w['end_bkjd']
        inside=(t>start)&(t<end)
        count=int(inside.sum())
        original=int(((raw_t>start)&(raw_t<end)&np.isfinite(raw_t)).sum())
        li=int(np.searchsorted(t,start,side='right'))-1
        ri=int(np.searchsorted(t,end,side='left'))
        bracketed=li>=0 and ri<len(t)
        maximum_gap=float(np.diff(t[li:ri+1]).max()) if bracketed and ri>li else None
        near_sides=bracketed and start-t[li]<=rule['maximum_gap_in_cadences']*cadence and t[ri]-end<=rule['maximum_gap_in_cadences']*cadence
        continuous=count>=rule['minimum_inside_points'] and near_sides and maximum_gap is not None and maximum_gap<=rule['maximum_gap_in_cadences']*cadence
        if count==0:
            status='窗内没有有效点'
        elif continuous:
            status='跨窗连续采样；信号可测性待判断'
        else:
            status='时间采样不完整；信号可测性待判断'
        reason=f'原始窗内 {original} 点，保留 {count} 点；两侧近邻={bool(near_sides)}；最大相邻间隔={maximum_gap} 天。'
        plot_name=f'M26-comparison-v1/events/cycle-{w["cycle"]}.png'
        local=(t>start-2*frozen['ephemeris']['duration_day'])&(t<end+2*frozen['ephemeris']['duration_day'])
        fig,ax=plt.subplots(figsize=(7,3))
        ax.errorbar(t[local],(holdout_flat.flux.value[local]-1)*1e6,
            yerr=holdout_flat.flux_err.value[local]*1e6,fmt='.',ms=4,alpha=.7)
        ax.axvspan(start,end,color='#e19a58',alpha=.25)
        ax.set(xlabel='BKJD (day)',ylabel='Relative flux - 1 (ppm)',title=f'Frozen event {w["cycle"]}: {count} accepted inside points')
        fig.tight_layout();save_plot(plot_name);plt.close(fig)
        windows_rows.append({'cycle':w['cycle'],'predicted_centre_bkjd':w['centre_bkjd'],
            'start_bkjd':start,'end_bkjd':end,'inside_count':count,'raw_inside_count':original,
            'coverage_status':status,'reason':reason,'maximum_gap_day':maximum_gap,
            'first_json_inside_count':given_counts.get(w['cycle'],''),
            'first_json_row_found':w['cycle'] in given_counts,'local_plot':plot_name,
            'learner_judgement':'','judgement_evidence':''})
    assert [r['cycle'] for r in windows_rows]==[r['cycle'] for r in forecasts]
    save_csv('M26-comparison-v1/windows.csv',list(windows_rows[0]),windows_rows)
    q3=read_json(RUN/'audit/development-results.json')['fixed_box_depth']
    q4=first_holdout['fixed_box_depth']
    aggregate={'definition':'Q4 minus Q3; whole-segment fixed-box estimates, not event-by-event fits',
        'q3':q3,'q4_first':q4,'depth_difference_ppm':q4['depth_ppm']-q3['depth_ppm'],
        'combined_empirical_scale_ppm':float(np.hypot(q3['empirical_standard_error_ppm'],q4['empirical_standard_error_ppm'])),
        'descriptive_depth_difference_ratio':first_holdout['depth_difference_combined_empirical_sigma'],
        'timing_diagnostic':first_holdout['timing_diagnostic'],'criteria':first_holdout['criteria'],
        'checks':first_holdout['checks'],
        'sampling_counts':{s:sum(r['coverage_status']==s for r in windows_rows) for s in sorted({r['coverage_status'] for r in windows_rows})},
        'sources':{'q3':'audit/development-results.json#fixed_box_depth',
            'q4':'audit/holdout-first-results.json#fixed_box_depth',
            'predictions':'audit/holdout-predictions-before-reveal.json#windows'},
        'single_event_detections':'not measured; do not infer from counts',
        'caveat':'Descriptive checks do not give a calibrated discovery probability.'}
    save_json('M26-comparison-v1/aggregate.json',aggregate)
    keep_text('M26-comparison-v1/comparison.md','# 本人比较\n\n[待填写] 一个预测编号的采样事实、图表依据、仍不确定的内容。\n\n[待填写] 整体深度差及相减顺序：\n\n[待填写] 为什么整体移位不是每个事件的实际中心：\n')
    print('原预测/输出行数：',len(forecasts),len(windows_rows))
    print('Q4−Q3 深度差 ppm：',aggregate['depth_difference_ppm'])
    print('时间采样分组：',aggregate['sampling_counts'])
    print('这些采样标签没有宣布逐个事件检出。请完成本人证据判断。')
````
